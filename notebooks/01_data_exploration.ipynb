```python
# Materials Property Prediction - Data Exploration

This notebook explores the materials data and performs initial analysis.
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette('husl')

%matplotlib inline

print("✓ Imports successful!")
```

```python
## 1. Load Data
```

```python
# Load band gap data
df = pd.read_csv('../data/band_gaps.csv')

print(f"Dataset shape: {df.shape}")
print(f"\nColumns: {df.columns.tolist()}")
df.head()
```

```python
## 2. Basic Statistics
```

```python
# Summary statistics
df.describe()
```

```python
# Check for missing values
missing = df.isnull().sum()
print("Missing values:")
print(missing[missing > 0])
```

```python
## 3. Band Gap Distribution
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Full distribution
axes[0].hist(df['band_gap'], bins=50, edgecolor='black', alpha=0.7)
axes[0].set_xlabel('Band Gap (eV)')
axes[0].set_ylabel('Count')
axes[0].set_title('Band Gap Distribution (All Materials)')
axes[0].axvline(df['band_gap'].mean(), color='red', linestyle='--', 
                label=f'Mean: {df["band_gap"].mean():.2f} eV')
axes[0].legend()

# Distribution for materials with band gap > 0
semiconductors = df[df['band_gap'] > 0]
axes[1].hist(semiconductors['band_gap'], bins=50, edgecolor='black', 
             alpha=0.7, color='orange')
axes[1].set_xlabel('Band Gap (eV)')
axes[1].set_ylabel('Count')
axes[1].set_title('Band Gap Distribution (Semiconductors & Insulators)')
axes[1].axvline(semiconductors['band_gap'].mean(), color='red', 
                linestyle='--', label=f'Mean: {semiconductors["band_gap"].mean():.2f} eV')
axes[1].legend()

plt.tight_layout()
plt.show()

print(f"\nTotal materials: {len(df)}")
print(f"Conductors (band gap = 0): {(df['band_gap'] == 0).sum()}")
print(f"Semiconductors/Insulators: {(df['band_gap'] > 0).sum()}")
```
